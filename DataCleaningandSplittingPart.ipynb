2. Handling null values

```python
#Exploring the unique values of the column (Loan Date End)
df_fifa21['Loan Date End'].unique().tolist()
```

From the exploration of this column, most players are not in loan and players who have a loan are permanent players in his club so, we will fill nulls with 
'Unknown' so that, we can define the analysis on Non-loan and loan players

```python
#Filling null values of this column with 'Unknown'
df_fifa21['Loan Date End'] = df_fifa21['Loan Date End'].fillna('Unknown')

#Final Check 
df_fifa21['Loan Date End'].isnull().sum()
```

```python
#Describe the unique values of the 'Hits' column
df_fifa21['Hits'].unique().tolist()
```

This column has mixed datatypes (int, float and str) and there are some null values in this columns so, before the type conversions we will fill nulls with '0.0' (string)

```python
#Filling null values of this column with 'Unknown'
df_fifa21['Hits'] = df_fifa21['Hits'].fillna('0.0')

#Final Check 
df_fifa21['Hits'].isnull().sum()
```

3. Type Conversions

```python
#Discovering the values of the first row in this dataset
pd.set_option('display.max_rows', None)
df_fifa21.iloc[0, :]
```

```python
#Converting the columns (Height and Weight) from str into float
df_fifa21['Height'] = df_fifa21['Height'].astype(str).str.extract(r'(\d+)').astype(float)
df_fifa21['Weight'] = df_fifa21['Weight'].astype(str).str.extract(r'(\d+)').astype(float)

#Checking
df_fifa21[['Height', 'Weight']].dtypes
```

```python
#Converting the columns (Value, Wage, and Release Clause) from str into float
df_fifa21['Value'] = df_fifa21['Value'].astype(str).str.extract(r'(\d+)').astype(float)
df_fifa21['Wage'] = df_fifa21['Wage'].astype(str).str.extract(r'(\d+)').astype(float)
df_fifa21['Release Clause'] = df_fifa21['Release Clause'].astype(str).str.extract(r'(\d+)').astype(float)

#Checking
df_fifa21[['Value', 'Wage', 'Release Clause']].dtypes
```

```python
#Working with hits column
#Cinverting ..k values into the corresponding number
has_k = df_fifa21['Hits'].str.upper().astype(str).str.endswith('K').values.tolist()
has_k
```

```python
#Stripping 'K' sign from some values in a column
df_fifa21['Hits'] = df_fifa21['Hits'].fillna(0).astype(str).str.replace('K', '').astype(float)
```

```python
#Multiplying values that had 'K' values
df_fifa21.loc[has_k, 'Hits'] *= 1000
```

6. Data Splitting

```python
#Defining unique names of players
print(len(df_fifa21['Name'].unique().tolist()))
```

```python
#Extract unique player names (keeps the first occurrence of each name)
dim_player = (
    df_fifa21[["Name", "Preferred Foot", "Attacking Work Rate", "Defensive Work Rate"]]
    .drop_duplicates(subset=["Name"])
    .reset_index(drop=True)
)

#Insert sequential PlayerID starting from 1
dim_player.insert(0, "PlayerID", range(1, len(dim_player) + 1))

#Rename columns to match your schema
dim_player.columns = [
    "PlayerID",
    "PlayerName",
    "PreferredFoot",
    "AttackingWorkRate",
    "DefensiveWorkRate",
]

dim_player.head()
```

```python
#Defining the number of distinct clubs 
print(len(df_fifa21["Club"].unique().tolist()))
```

```python
#Creating a new column of Dim CLub
dim_club = pd.DataFrame({
    "ClubID": range(1,  len(df_fifa21["Club"].unique().tolist()) + 1),
    "Club": df_fifa21["Club"].unique()
})

dim_club
```

```python
#Defining the number of unique nationalities
print(len(df_fifa21["Nationality"].unique().tolist()))
```

```python
#Define the column Dim nationality
dim_nationality = pd.DataFrame({
    "NationalityID": range(1, len(df_fifa21["Nationality"].unique().tolist()) + 1),
    "Nationality": df_fifa21["Nationality"].unique()
})
dim_nationality
```

```python
# Extract unique combinations of Best Position and Positions
dim_position = (
    df_fifa21[["Best Position", "Positions"]]
    .drop_duplicates()
    .reset_index(drop=True)
)

# Add surrogate primary key Position_ID starting from 1
dim_position.insert(0, "Position_ID", range(1, len(dim_position) + 1))

dim_position
```

```python
#Create Dim_Contract
dim_contract = (
    df_fifa21[["Contract", "Loan Date End"]]
    .drop_duplicates()
    .reset_index(drop=True)
)
dim_contract.insert(0, "Contract_ID", range(1, len(dim_contract) + 1))

dim_contract.head()
```

```python
#Create Dim_Date
dim_date = (
    df_fifa21[["Joined", "Joined Year", "Joined Month"]]
    .drop_duplicates()
    .reset_index(drop=True)
)
dim_date.insert(0, "Date_ID", range(1, len(dim_date) + 1))

dim_date.head()
```

```python
#Initializing the fact player table
fact_player = (
    df_fifa21
    .merge(dim_player[["PlayerID", "PlayerName"]], left_on="Name", right_on="PlayerName", how="left")
    .merge(dim_club, on="Club", how="left")
    .merge(dim_nationality, on="Nationality", how="left")
    .merge(dim_position, on=["Best Position", "Positions"], how="left")
    .merge(dim_contract, on=["Contract", "Loan Date End"], how="left")
    .merge(dim_date, on=["Joined", "Joined Year", "Joined Month"], how="left")
)
```

```python
#Dropping redundant columns in fact table
cols_to_drop = [
    "Name", "PlayerName", "Preferred Foot", "Attacking Work Rate", "Defensive Work Rate",
    "Club",
    "Nationality",
    "Best Position", "Positions",
    "Contract", "Loan Date End",
    "Joined", "Joined Year", "Joined Month"
]
fact_player = fact_player.drop(columns=cols_to_drop)
```

```python
#Moving Primary Key and Foreign Keys to the front of the Fact table
fk_cols = ["ID", "PlayerID", "ClubID", "NationalityID", "Position_ID", "Contract_ID", "Date_ID"]
metric_cols = [col for col in fact_player.columns if col not in fk_cols]
fact_player = fact_player[fk_cols + metric_cols]
```

```python
#Verify row count stayed identical
assert len(fact_player) == len(df_fifa21), "Row count mismatch after merging!"
print(f"Fact table shape: {fact_player.shape}")
fact_player.head()
```

```python
import os

# Create an output folder in your current directory
output_dir = "fifa21_star_schema"
os.makedirs(output_dir, exist_ok=True)

# Map each CSV filename to its DataFrame
tables = {
    "fact_player.csv": fact_player,
    "dim_player.csv": dim_player,
    "dim_club.csv": dim_club,
    "dim_nationality.csv": dim_nationality,
    "dim_position.csv": dim_position,
    "dim_contract.csv": dim_contract,
    "dim_date.csv": dim_date,
}

# Export all tables to CSV
for filename, df in tables.items():
    filepath = os.path.join(output_dir, filename)
    df.to_csv(filepath, index=False, encoding="utf-8-sig")
    print(f"Saved {filename} -> {df.shape[0]} rows, {df.shape[1]} columns")
```
